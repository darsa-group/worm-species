```python
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
data = '/home/devd/erda-home/Earthworms/petridish-worm-images'

global_metadata = pd.read_csv(os.path.join(data,'01_Segmented','global_metadata.csv'))


global_metadata['species'] = global_metadata['barcode'].str.replace(r'_\d+$', '', regex=True)
global_metadata['date'] = pd.to_datetime(global_metadata['timestamp']).apply(lambda x: x.date())
global_metadata[["taxon_label", "life_stage"]] = (
    global_metadata["species"]
    .str.extract(r"^(.+)_(Adult|Juvenile)$")
)
global_metadata["genus"] = global_metadata["taxon_label"].str.split("_").str[0]
global_metadata["species_label"] = global_metadata["taxon_label"]
```

```python
global_metadata
```

```python
#remove number at the end of barcode

#make grpahs about species distribution
plt.figure(figsize=(10,5))
global_metadata['species'].value_counts().plot(kind='bar')
plt.title('Species Distribution')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()

#unique individuals per species
unique_individuals = global_metadata.groupby('species')['barcode'].nunique()
unique_individuals.plot(kind='bar', figsize=(10,5))
plt.title('Unique Individuals per Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()


#also split by date and make a graph of species distribution over time
plt.figure(figsize=(25,15))
global_metadata.groupby(['date','species']).size().unstack().plot(kind='bar',stacked=True, figsize=(25,15))
plt.title('Species Distribution Over Time')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend(title='Species', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()



plt.figure(figsize=(10,5))
global_metadata['life_stage'].value_counts().plot(kind='bar')
plt.title('Life Stage Distribution')
plt.xlabel('Life Stage')
plt.ylabel('Count')

plt.show()

#show distribution of species by life stage unique individuals
plt.figure(figsize=(10,5))
global_metadata.groupby(['species_label','life_stage'])['barcode'].nunique().unstack().plot(kind='bar', stacked=True, figsize=(10,5))
plt.title('Unique Individuals per Species by Life Stage')
plt.xlabel('Species')
plt.ylabel('Count')
plt.yscale('log')
plt.show()

#unique individuals per genus adult vs juvenile
plt.figure(figsize=(10,5))
global_metadata.groupby(['genus','life_stage'])['barcode'].nunique().unstack().plot(kind='bar', stacked=True, figsize=(10,5))
plt.title('Unique Individuals per Genus by Life Stage')
plt.xlabel('Genus')
plt.ylabel('Count')
plt.show()


```

```python
plt.figure(figsize=(8,5))
global_metadata["weight_g"].hist(bins=40)
plt.xlabel("Weight (g)")
plt.ylabel("Count")
plt.title("Earthworm Weight Distribution")
plt.show()


plt.figure(figsize=(12,5))
global_metadata.boxplot(column="weight_g", by="species_label", rot=90)
plt.suptitle("")
plt.title("Weight Distribution by Species")
plt.ylabel("Weight (g)")
plt.show()

plt.figure(figsize=(6,5))
global_metadata.boxplot(column="weight_g", by="life_stage")
plt.suptitle("")
plt.title("Weight by Life Stage")
plt.ylabel("Weight (g)")
plt.show()

counts = global_metadata.groupby("species_label")["barcode"].nunique()

counts.sort_values(ascending=False).plot(
    kind="bar",
    logy=True,
    figsize=(12,5)
)

plt.ylabel("Unique Individuals (log scale)")
plt.title("Class Imbalance")
plt.show()
```

```python
#show 5 random images and segmented images from each species
import random
from PIL import Image


for species in global_metadata['species'].unique():
    species_data = global_metadata[global_metadata['species'] == species]
    random_images = species_data.sample(n=5, random_state=54)
    plt.figure(figsize=(15,5))
    for i, row in enumerate(random_images.itertuples()):
        raw_image_path = os.path.join(data, row.rel_path_raw )
        segmented_image_path = os.path.join(data,  row.rel_path_seg )
        raw_image = Image.open(raw_image_path)
        segmented_image = Image.open(segmented_image_path)
        plt.subplot(2, 5, i + 1)
        plt.imshow(raw_image)
        plt.title(f'Raw: {row.barcode}')
        plt.axis('off')
        plt.subplot(2, 5, i + 6)
        plt.imshow(segmented_image)
        plt.title(f'Segmented: {row.barcode}')
        
    plt.show()
```
