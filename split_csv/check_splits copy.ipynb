```python
import pandas as pd
from pathlib import Path

base = Path("/home/devd/worm-species/split_csv")

train = pd.read_csv(base / "train_split.csv")
val = pd.read_csv(base / "val_split.csv")
test = pd.read_csv(base / "test_split.csv")

group_col = "barcode"
image_cols = ["filename", "rel_path_raw", "rel_path_seg", "rel_path_segmask"]
target_col = "species_label"

def overlap(a, b, col):
    if col not in a.columns or col not in b.columns:
        return set()
    return set(a[col].dropna().astype(str)) & set(b[col].dropna().astype(str))

for col in [group_col, target_col, *image_cols]:
    print(f"\n=== {col} ===")
    for n1, d1, n2, d2 in [
        ("train", train, "val", val),
        ("train", train, "test", test),
        ("val", val, "test", test),
    ]:
        inter = overlap(d1, d2, col)
        print(f"{n1} vs {n2}: {len(inter)}")
        if inter:
            print(sorted(list(inter))[:10])

for name, df in [("train", train), ("val", val), ("test", test)]:
    print(f"\n{name}: rows={len(df)} unique_groups={df[group_col].nunique() if group_col in df else 'NA'}")
    if target_col in df.columns:
        print(df[target_col].value_counts().head(10))
```
